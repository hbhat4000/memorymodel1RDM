```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
coeffs = np.loadtxt('./pinnresults/fci_heh+_6-31gss_0.025000_100_coeffs.txt',
                    delimiter=',',dtype=np.complex128)
```

```python
for j in range(100):
    plt.figure(figsize=(3,2))
    plt.plot(coeffs[:1000,j].real)
    plt.plot(coeffs[:1000,j].imag)
    plt.show()
```
